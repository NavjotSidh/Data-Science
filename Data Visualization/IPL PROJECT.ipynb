```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
df=pd.read_csv(r"C:\NAVJOT\PYCHARM\CSV\Ipl 2026\IPL 2022.csv")
df.columns
```

```python
# Which team won most matches ?
match_win=df['match_winner'].value_counts()
sns.barplot(y=match_win.index,x=match_win.values,palette='rainbow')
plt.title("Match won by each team")
```

```python
# Toss decision
sns.countplot(x=df['toss_decision'])
plt.title("Toss Decision Trend")
plt.show()
```

```python
# Toss Winner vs Match Winner
count=df[df["toss_winner"]==df["match_winner"]]["match_winner"].count()
percent=(count*100)/df.shape[0]
print(percent)
```

```python
# How team won , Runs vs Wickets
sns.countplot(x=df['won_by'])
```

Key Player Performance

```python
count=df['player_of_the_match'].value_counts().head(10)
sns.barplot(y=count.index,x=count.values,palette="rainbow")
plt.title("Most Player of The match")
```
